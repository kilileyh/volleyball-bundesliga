```python
import pandas as pd

matches = pd.read_csv("../data/processed/matches.csv")
matchstats = pd.read_csv("../data/processed/matchstats.csv")
matches.head()
```

```python
match_keys = ["season_id", "league_gender", "match_id"]
team_keys = match_keys + ["team_id"]
```

```python
matches.duplicated(team_keys).sum()
```

```python
matches.duplicated(["match_id"]).sum()
```

```python
matches.groupby(match_keys).size().value_counts()
```

```python
matches["win"] = (matches["set_won"] == 3).astype(int)
matches.groupby(match_keys)["win"].sum().value_counts()
```

```python

player_keys = team_keys + ["player_id"]
matchstats[matchstats.duplicated(player_keys, keep=False)]

# 4 duplicated player rows → remove before aggregating.
```

```python
teams_with_stats = matchstats[team_keys].drop_duplicates()
check = matches.merge(teams_with_stats, on=team_keys, how="left", indicator=True)
check["_merge"].value_counts()

# 8 matches have no player statistics → excluded for both teams.
```
